```python
import pandas as pd
import numpy as np

df_vendas = pd.read_csv("vendas_tech.csv", low_memory=False)
display(df_vendas.info())
display(df_vendas.columns)
display(df_vendas.shape)
# display(df_vendas.head(20))
# display(df_vendas.describe())
```

# Abaixo vou colocar todos os comandos para tratar os dados:

```python
#Visualizando os dados da colunas "Loja" e "Cliente"
display(df_vendas[["Loja", "Cliente"]])
```

```python
#Como excluir colunas, nesse caso vamos excluir a coluna "Data_Base" pois ela esta inteira vazia:
df_analise = df_vendas.drop(columns=["Data_Base"])
display(df_analise)
```

```python
#tem como excluir todas as linhas que tem algum valor vazio:
#se voce reparar no resultado da "df_vendas.info()" a coluna "Loja" tinha a mesma quantidade de linhas que ficou agora depois do "dropna"
df_analise = df_analise.dropna()
display(df_analise)
```

```python
#aqui vou preencher os valores vazios da coluna "Loja" com a informação "Online"
df_analise = df_vendas.drop(columns=["Data_Base"])
df_analise["Loja"] = df_analise["Loja"].fillna("Online")
display(df_analise)
```

```python
#aqui vamos alterar as datas para o formato (datetime64[us]) pois o pandas tinha reconhecido eles como texto(object)
#para isso usamos "pd.to_datetime" e temos que passar o formato da data:
#para achar como digitar o formato usar os links abaixo:

#https://pandas.pydata.org/docs/reference/api/pandas.to_datetime.html
#https://docs.python.org/3/library/datetime.html#strftime-and-strptime-behavior

df_analise["Data"] = pd.to_datetime(df_analise["Data"], format="%Y-%m-%d")
display(df_analise)
display(df_analise.info())
```
